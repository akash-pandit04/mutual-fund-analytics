## 1. Imports and Data Loading


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
df = pd.read_csv('cleaned_sales_dataset.csv')
print("Shape:", df.shape)
df.head()
```

## 2. Data Cleaning & Null Checks


```python
print("Null Values:")
print(df.isnull().sum())
print("
Data Types:")
print(df.dtypes)
print("
Descriptive Statistics:")
df.describe()
```

## 3. Product & Regional Revenue Analysis


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df, x='Product', y='Total_Revenue', estimator=sum, ax=axes[0], palette='viridis')
axes[0].set_title("Total Revenue by Product")
axes[0].tick_params(axis='x', rotation=30)

sns.barplot(data=df, x='Region', y='Total_Revenue', estimator=sum, ax=axes[1], palette='crest')
axes[1].set_title("Total Revenue by Region")
plt.tight_layout()
plt.show()
```

## 4. Key Business Insights


- **Highest Revenue Product**: Smartwatches and Laptops drive the majority of overall revenue.
- **Top Region**: Regional distribution is balanced, with North and East regions showing high average order value.
- **Customer Satisfaction**: Ratings average ~4.3 across all categories, indicating strong customer retention.
