## 1. Setup & Analytics Engine Invocation


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3

conn = sqlite3.connect('../data/processed/database.db')
risk_df = pd.read_sql("SELECT * FROM risk_metrics", conn)
print("Loaded Risk Metrics for", len(risk_df), "schemes.")
risk_df.head()
```

## 2. Alpha & Beta Distribution


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(risk_df['alpha'], kde=True, ax=axes[0], color='teal')
axes[0].set_title("Alpha (Excess Return) Distribution")

sns.histplot(risk_df['beta'], kde=True, ax=axes[1], color='coral')
axes[1].set_title("Beta (Market Sensitivity) Distribution")
plt.tight_layout()
plt.show()
```

## 3. K-Means Risk-Return Profile Clustering


```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=risk_df, x='annualized_std_dev', y='alpha', hue='risk_profile', style='risk_profile', s=100, palette={'Low Risk': 'green', 'Medium Risk': 'orange', 'High Risk': 'red'})
plt.title("K-Means Risk Profiles (k=3)")
plt.xlabel("Annualized Standard Deviation (Risk)")
plt.ylabel("Alpha (Excess Return)")
plt.axhline(0, color='gray', linestyle='--')
plt.show()
```

## 4. SMA / EMA Moving Average Trend Signals


```python
nav_df = pd.read_sql("SELECT * FROM nav_history WHERE amfi_code = 125497", conn)
nav_df['date'] = pd.to_datetime(nav_df['date'], dayfirst=True)
nav_df = nav_df.sort_values('date').tail(365)

nav_df['sma_30'] = nav_df['nav'].rolling(30).mean()
nav_df['ema_30'] = nav_df['nav'].ewm(span=30, adjust=False).mean()

plt.figure(figsize=(12, 5))
plt.plot(nav_df['date'], nav_df['nav'], label='NAV', color='gray', alpha=0.5)
plt.plot(nav_df['date'], nav_df['sma_30'], label='30-Day SMA', color='blue')
plt.plot(nav_df['date'], nav_df['ema_30'], label='30-Day EMA', color='magenta', linestyle=':')
plt.title("NAV Trend & Moving Averages (Scheme 125497)")
plt.legend()
plt.show()
```
